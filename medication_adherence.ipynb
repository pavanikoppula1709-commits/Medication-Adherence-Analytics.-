```python
df["Age_Group"] = pd.cut(
    df["Age"],
    bins=[17, 30, 45, 60, 90],
    labels=["18-30", "31-45", "46-60", "61+"]
)
```

```python
df["Adherence_Status"] = df["Adherence"].map({
    1: "Adherent",
    0: "Non-Adherent"
})
```

```python
df["Insurance_Status"] = df["Insurance_Coverage"].map({
    1: "Covered",
    0: "Not Covered"
})
```

```python
df["Previous_Adherence_Status"] = df["Previous_Adherence"].map({
    1: "Previously Adherent",
    0: "Previously Non-Adherent"
})
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="Adherence_Status"
)

plt.title("Medication Adherence Distribution")
plt.xlabel("Adherence Status")
plt.ylabel("Number of Patients")

plt.show()
```

```python
age_adherence = (
    df.groupby("Age_Group", observed=True)["Adherence"]
      .mean()
      .mul(100)
      .reset_index()
)

plt.figure(figsize=(8, 5))

sns.barplot(
    data=age_adherence,
    x="Age_Group",
    y="Adherence"
)

plt.title("Medication Adherence Rate by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Adherence Rate (%)")

plt.ylim(0, 100)

plt.show()
```

```python
gender_adherence = (
    df.groupby("Gender")["Adherence"]
      .mean()
      .mul(100)
      .reset_index()
)

plt.figure(figsize=(7, 5))

sns.barplot(
    data=gender_adherence,
    x="Gender",
    y="Adherence"
)

plt.title("Medication Adherence Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Adherence Rate (%)")

plt.ylim(0, 100)

plt.show()
```

```python
med_adherence = (
    df.groupby("Medication_Type")["Adherence"]
      .mean()
      .mul(100)
      .reset_index()
    .sort_values("Adherence", ascending=False)
)

plt.figure(figsize=(8, 5))

sns.barplot(
    data=med_adherence,
    x="Medication_Type",
    y="Adherence"
)

plt.title("Medication Adherence Rate by Medication Type")
plt.xlabel("Medication Type")
plt.ylabel("Adherence Rate (%)")

plt.ylim(0, 100)

plt.show()
```
